## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
import os

# Define the repository name and path
repo_name = "FlyRank1"
repo_url = "https://github.com/Zeyneb00/FlyRank1.git"
repo_path = f"/content/{repo_name}" # Standard Colab cloning location

# 1. Ensure the repository is cloned and change directory to its root
if not os.path.exists(repo_path):
    print(f"Cloning {repo_name} repository...")
    os.system(f"git clone {repo_url} {repo_path}")
    print("Repository cloned.")
else:
    print(f"{repo_name} repository already exists.")

# Change current working directory to the repository root
os.chdir(repo_path)
print(f"Changed working directory to: {os.getcwd()}")

# 2. Load the data file from the repo (now relative to repo root)
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# 3. Calculate 3 simple numbers
total_pages = len(df)
avg_views = df["impressions_90d"].mean()
declining_pages_pct = (df["trend_direction"] == "down").mean() * 100

print(f"\nNumber 1: Total pages analyzed = {total_pages:,}")
print(f"Number 2: Average 90-day search views per page = {avg_views:,.0f}")
print(f"Number 3: Percentage of pages declining in traffic = {declining_pages_pct:.1f}%")
```
